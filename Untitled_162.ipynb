##Step 9: Cantatenating description and designation, and filling nan in description with empty strings

cancatening because designation is short, description contains additional semantic information, missing descriptions might become empty strings.

```python
import re
import unidecode

FRENCH_STOPWORDS = set("""
au aux avec ce ces dans de des du elle en et eux il je la le leur lui ma mais me
meme mes moi mon ne nos notre nous on ou par pas pour qu que qui sa se ses son sur
ta te tes toi ton tu un une vos votre vous c d j l a m n s t y ete etee etees etes
etant suis es est sommes etes sont serai seras sera serons serez seront les des
""".split())

def clean_french_text(text):

    if pd.isna(text):
        return ""
    text = str(text)
    text = re.sub(r'<[^>]+>', ' ', text)
    text = unidecode.unidecode(text)
    text = text.lower()
    text = re.sub(r'http\S+|www\.\S+', ' ', text)
    text = re.sub(r'[^a-z0-9\s]', ' ', text)
    text = re.sub(r'\d+', ' ', text)
    tokens = [t for t in text.split() if t not in FRENCH_STOPWORDS and len(t) > 1]
    return ' '.join(tokens)

df["description"] = df["description"].fillna("")

df["designation_clean"] = df["designation"].apply(clean_french_text)
df["description_clean"] = df["description"].apply(clean_french_text)

df["text"] = (df["designation_clean"] + " " + df["description_clean"]).str.strip()
df["text"] = df["text"].replace("", "produit sans description")

print(f"Empty-after-cleaning rows backfilled: {(df['text'] == 'produit sans description').sum()}")
df[["designation_clean", "description_clean", "text"]].head()
```

## Step 17: Approach B: Multimodal Fusion Network (MobileNetV2 + TF-IDF/SVD text)

Builds on the classical baseline above: reuses the exact same stratified split
(x_train/x_test/y_train/y_test) and the fitted TF-IDF matrices
(x_train_tfidf/x_test_tfidf) so Approach A and Approach B are compared on
identical held-out rows. Text is compressed to a small dense vector via
TruncatedSVD to keep the ONNX-exported model small and fast, this is the
architecture chosen for its favorable CPU latency profile against the
30ms production budget

```python
from sklearn.preprocessing import LabelEncoder

nn_train_df = df.loc[x_train.index].copy()
nn_val_df   = df.loc[x_test.index].copy()

label_encoder = LabelEncoder()
label_encoder.fit(df["prdtypecode"])
nn_train_df["label_encoded"] = label_encoder.transform(nn_train_df["prdtypecode"])
nn_val_df["label_encoded"]   = label_encoder.transform(nn_val_df["prdtypecode"])

NUM_CLASSES = len(label_encoder.classes_)
print(f"NUM_CLASSES = {NUM_CLASSES}")
print(f"Train rows = {len(nn_train_df)} | Val rows = {len(nn_val_df)}")
```

```python
from sklearn.decomposition import TruncatedSVD

TEXT_EMBED_DIM = 300

svd = TruncatedSVD(n_components=TEXT_EMBED_DIM, random_state=SEED)
train_text_dense = svd.fit_transform(x_train_tfidf).astype(np.float32)
val_text_dense   = svd.transform(x_test_tfidf).astype(np.float32)

print(f"Text embedding dim: {TEXT_EMBED_DIM} | Explained variance: {svd.explained_variance_ratio_.sum():.3f}")

nn_train_df = nn_train_df.reset_index(drop=True)
nn_val_df   = nn_val_df.reset_index(drop=True)
```

### Step 18: Class imbalance handling for the neural network (weights + sampler)

```python
from sklearn.utils.class_weight import compute_class_weight
from torch.utils.data import WeightedRandomSampler

class_weights_arr = compute_class_weight(
    class_weight="balanced",
    classes=np.arange(NUM_CLASSES),
    y=nn_train_df["label_encoded"].values,
)
class_weights_tensor = torch.tensor(class_weights_arr, dtype=torch.float32).to(dev)

sample_weights = class_weights_arr[nn_train_df["label_encoded"].values]
sampler = WeightedRandomSampler(weights=sample_weights, num_samples=len(sample_weights), replacement=True)
```

### Step 19: Multimodal Dataset using Pytorch (resilient image loader)

```python
from pathlib import Path
from typing import Optional
from PIL import Image, UnidentifiedImageError
from torch.utils.data import Dataset
import torchvision.transforms as T

IMAGE_SIZE = 224
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

train_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.RandomHorizontalFlip(p=0.5),
    T.ColorJitter(brightness=0.15, contrast=0.15),
    T.ToTensor(),
    T.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])
eval_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.ToTensor(),
    T.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])


class MultimodalProductDataset(Dataset):

    PLACEHOLDER_COLOR = (128, 128, 128)

    def __init__(self, dataframe, image_dir, text_vectors, transform=None):
        self.df = dataframe.reset_index(drop=True)
        self.image_dir = Path(image_dir)
        self.text_vectors = text_vectors
        self.transform = transform
        self.n_missing = 0
        self.n_corrupted = 0

    def __len__(self):
        return len(self.df)

    def _resolve_path(self, row) -> Optional[Path]:
        try:
            return self.image_dir / f"image_{int(row['imageid'])}_product_{int(row['productid'])}.jpg"
        except (KeyError, ValueError, TypeError):
            return None

    def _load_image(self, image_path: Optional[Path]) -> Image.Image:
        if image_path is None or not image_path.exists():
            self.n_missing += 1
            return Image.new("RGB", (IMAGE_SIZE, IMAGE_SIZE), color=self.PLACEHOLDER_COLOR)
        try:
            with Image.open(image_path) as im:
                im.verify()
            # verify() closes the file handle, it must reopen to really decode pixels
            return Image.open(image_path).convert("RGB")
        except (UnidentifiedImageError, OSError, ValueError):
            self.n_corrupted += 1
            return Image.new("RGB", (IMAGE_SIZE, IMAGE_SIZE), color=self.PLACEHOLDER_COLOR)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img = self._load_image(self._resolve_path(row))
        if self.transform:
            img = self.transform(img)
        text_vec = torch.as_tensor(self.text_vectors[idx], dtype=torch.float32)
        label = int(row["label_encoded"])
        return img, text_vec, label

    def failure_report(self):
        total = len(self.df)
        return {
            "total": total,
            "missing": self.n_missing,
            "corrupted": self.n_corrupted,
            "failure_rate_pct": round((self.n_missing + self.n_corrupted) / total * 100, 2),
        }
```

```python
from torch.utils.data import DataLoader

BATCH_SIZE = 64
NUM_WORKERS = 2

train_dataset = MultimodalProductDataset(nn_train_df, train_image_dir, train_text_dense, transform=train_transform)
val_dataset   = MultimodalProductDataset(nn_val_df, train_image_dir, val_text_dense, transform=eval_transform)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, sampler=sampler,
                           num_workers=NUM_WORKERS, pin_memory=(dev.type == "cuda"))
val_loader   = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False,
                           num_workers=NUM_WORKERS, pin_memory=(dev.type == "cuda"))

print(f"Train batches: {len(train_loader)} | Val batches: {len(val_loader)}")
```

### Step 20: Multimodal fusion model MobileNetV2 (frozen) + text MLP

MobileNetV2 is chosen over ResNet18 specifically to protect the CPU latency
budget. Also dataset is not very huge, RestNet will fail on small datasets, it is suggested for 10,000+ images. We only have 500, MobileNetV2 is lightweight and feasible, and satisfies the budget and hardware limitations.

```python
import torch.nn as nn
import torchvision.models as tv_models

class MultimodalFusionNet(nn.Module):

    def __init__(self, num_classes: int, text_dim: int = 300, freeze_backbone: bool = True, dropout: float = 0.3):
        super().__init__()
        backbone = tv_models.mobilenet_v2(weights=tv_models.MobileNet_V2_Weights.IMAGENET1K_V1)
        self.vision_backbone = backbone.features
        self.pool = nn.AdaptiveAvgPool2d(1)
        vision_out_dim = backbone.last_channel  # 1280

        if freeze_backbone:
            for p in self.vision_backbone.parameters():
                p.requires_grad = False

        self.vision_head = nn.Sequential(
            nn.Linear(vision_out_dim, 256), nn.ReLU(inplace=True), nn.Dropout(dropout)
        )
        self.text_head = nn.Sequential(
            nn.Linear(text_dim, 128), nn.ReLU(inplace=True), nn.Dropout(dropout)
        )
        self.classifier = nn.Sequential(
            nn.Linear(256 + 128, 128), nn.ReLU(inplace=True), nn.Dropout(dropout / 2),
            nn.Linear(128, num_classes)
        )

    def forward(self, image: torch.Tensor, text: torch.Tensor) -> torch.Tensor:
        v = self.vision_backbone(image)
        v = self.pool(v).flatten(1)
        v = self.vision_head(v)
        t = self.text_head(text)
        fused = torch.cat([v, t], dim=1)
        return self.classifier(fused)


TEXT_DIM = TEXT_EMBED_DIM
model_nn = MultimodalFusionNet(num_classes=NUM_CLASSES, text_dim=TEXT_DIM, freeze_backbone=True).to(dev)

n_trainable = sum(p.numel() for p in model_nn.parameters() if p.requires_grad)
n_total = sum(p.numel() for p in model_nn.parameters())
print(f"Trainable params: {n_trainable:,} / {n_total:,}")
```

### Step 21: Training loop (early stopping + best-checkpoint on macro F1 + grad clipping)

```python
from sklearn.metrics import f1_score, precision_score, recall_score
import copy

EPOCHS = 10
PATIENCE = 3
LR = 1e-3

criterion = nn.CrossEntropyLoss(weight=class_weights_tensor)
optimizer = torch.optim.Adam(filter(lambda p: p.requires_grad, model_nn.parameters()), lr=LR, weight_decay=1e-5)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="max", factor=0.5, patience=1)

best_f1 = -1.0
best_state = None
epochs_no_improve = 0

for epoch in range(1, EPOCHS + 1):
    model_nn.train()
    running_loss = 0.0
    for images, texts, labels in train_loader:
        images, texts, labels = images.to(dev), texts.to(dev), labels.to(dev)
        optimizer.zero_grad()
        logits = model_nn(images, texts)
        loss = criterion(logits, labels)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model_nn.parameters(), max_norm=5.0)
        optimizer.step()
        running_loss += loss.item() * images.size(0)
    train_loss = running_loss / len(train_dataset)

    model_nn.eval()
    val_preds, val_labels = [], []
    with torch.no_grad():
        for images, texts, labels in val_loader:
            images, texts = images.to(dev), texts.to(dev)
            preds = model_nn(images, texts).argmax(dim=1).cpu().numpy()
            val_preds.extend(preds)
            val_labels.extend(labels.numpy())

    val_f1 = f1_score(val_labels, val_preds, average="macro")
    scheduler.step(val_f1)
    print(f"Epoch {epoch:02d}/{EPOCHS} | train_loss={train_loss:.4f} | val_macro_f1={val_f1:.4f}")

    if val_f1 > best_f1:
        best_f1 = val_f1
        best_state = copy.deepcopy(model_nn.state_dict())
        epochs_no_improve = 0
    else:
        epochs_no_improve += 1
        if epochs_no_improve >= PATIENCE:
            print(f"Early stopping at epoch {epoch} (no improvement in {PATIENCE} epochs).")
            break

model_nn.load_state_dict(best_state)
print(f"\nBest validation macro F1 (Approach B): {best_f1:.4f}")
print("Image loader failure report (train):", train_dataset.failure_report())
print("Image loader failure report (val):  ", val_dataset.failure_report())
```

###Training Strategy

The model is trained using early stopping based on validation Macro F1.

The checkpoint with the highest validation Macro F1 is retained to avoid overfitting.

###Why ONNX?

### Step 24: PyTorch and ONNX parity check (catches silent export bugs before production)

```python
import onnxruntime as ort

def check_onnx_pytorch_parity(onnx_path, torch_model, n_samples=20, atol=1e-4):
    #compares ONNX Runtime outputs against PyTorch model

    sess = ort.InferenceSession(onnx_path, providers=["CPUExecutionProvider"])
    torch_model.eval()
    max_diff = 0.0
    with torch.no_grad():
        for _ in range(n_samples):
            img = torch.randn(1, 3, IMAGE_SIZE, IMAGE_SIZE)
            txt = torch.randn(1, TEXT_DIM)
            torch_out = torch_model(img, txt).numpy()
            onnx_out = sess.run(None, {"image": img.numpy(), "text_features": txt.numpy()})[0]
            max_diff = max(max_diff, np.abs(torch_out - onnx_out).max())
    status = "PASS" if max_diff < atol else "FAIL"
    print(f"[{status}] Max abs diff PyTorch vs ONNX: {max_diff:.6f} (tolerance={atol})")
    return max_diff < atol

parity_ok = check_onnx_pytorch_parity(ONNX_PATH, model_cpu)
assert parity_ok, "ONNX export does not numerically match the PyTorch model -- do not ship."
```

### Step 25: CPU latency benchmark (single thread, 1000 iterations, P50/P95/P99)

```python
import time

def benchmark_onnx_latency(onnx_path, text_dim, image_size=224, n_iters=1000, n_warmup=30, intra_threads=1):
    sess_options = ort.SessionOptions()
    sess_options.intra_op_num_threads = intra_threads
    sess_options.inter_op_num_threads = 1
    sess_options.execution_mode = ort.ExecutionMode.ORT_SEQUENTIAL
    sess_options.graph_optimization_level = ort.GraphOptimizationLevel.ORT_ENABLE_ALL

    session = ort.InferenceSession(onnx_path, sess_options=sess_options, providers=["CPUExecutionProvider"])

    feed = {
        "image": np.random.randn(1, 3, image_size, image_size).astype(np.float32),
        "text_features": np.random.randn(1, text_dim).astype(np.float32),
    }

    for _ in range(n_warmup):
        session.run(None, feed)

    latencies_ms = np.empty(n_iters, dtype=np.float64)
    for i in range(n_iters):
        t0 = time.perf_counter()
        session.run(None, feed)
        latencies_ms[i] = (time.perf_counter() - t0) * 1000.0
    return latencies_ms


latencies = benchmark_onnx_latency(ONNX_PATH, text_dim=TEXT_DIM)
p50, p95, p99 = np.percentile(latencies, [50, 95, 99])
print(f"[FP32] P50={p50:.2f}ms | P95={p95:.2f}ms | P99={p99:.2f}ms")
print(f"Meets <30ms P95 production budget: {'YES' if p95 < 30 else 'NO'}")
```

### Step 26: Conditional INT8 dynamic quantization + re benchmark

```python
from onnxruntime.quantization import quantize_dynamic, QuantType

QUANT_PATH = "multimodal_ecommerce_model_quant.onnx"
FINAL_ONNX_PATH, final_latencies, final_p50, final_p95, final_p99 = ONNX_PATH, latencies, p50, p95, p99

if p95 >= 30:
    print("P95 exceeds the 30ms budget -- applying dynamic INT8 quantization...")
    quantize_dynamic(ONNX_PATH, QUANT_PATH, weight_type=QuantType.QUInt8)

    fp32_mb = os.path.getsize(ONNX_PATH) / 1e6
    int8_mb = os.path.getsize(QUANT_PATH) / 1e6
    print(f"Size: FP32={fp32_mb:.2f}MB -> INT8={int8_mb:.2f}MB ({(1 - int8_mb/fp32_mb)*100:.0f}% smaller)")

    check_onnx_pytorch_parity(QUANT_PATH, model_cpu, atol=5e-2)  # quantization is lossy -> looser tolerance

    latencies_q = benchmark_onnx_latency(QUANT_PATH, text_dim=TEXT_DIM)
    p50_q, p95_q, p99_q = np.percentile(latencies_q, [50, 95, 99])
    print(f"[INT8] P50={p50_q:.2f}ms | P95={p95_q:.2f}ms | P99={p99_q:.2f}ms")
    print(f"Meets <30ms P95 budget after quantization: {'YES' if p95_q < 30 else 'NO'}")

    if p95_q < p95:
        FINAL_ONNX_PATH = QUANT_PATH
        final_latencies, final_p50, final_p95, final_p99 = latencies_q, p50_q, p95_q, p99_q
else:
    print("FP32 model already meets the 30ms P95 budget, quantization not required.")

print(f"\nFinal deployment artifact: {FINAL_ONNX_PATH}")
print(f"Final latency -- P50={final_p50:.2f}ms | P95={final_p95:.2f}ms | P99={final_p99:.2f}ms")
```
